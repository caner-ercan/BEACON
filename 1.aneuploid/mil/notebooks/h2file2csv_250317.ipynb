```python
import os
import h5py
import pandas as pd
import numpy as np
```

```python
def h52csv(h5_path,csv_path):
    with h5py.File(h5_path, 'r') as h5data:
        # A_minmax = np.array(h5data['attention_scores_minmax'])
        A_minmax = np.array(h5data['avg_attention_scores_minmax'])
        coords = np.array(h5data['coords'])
    
    df = pd.DataFrame({
        'x': coords[:, 0],
        'y': coords[:, 1],
        'A_minmax': A_minmax.flatten(),  # Flatten to ensure it's a 1D array
        # 'tile_probabilities': tile_probabilities.flatten()
    })
    df.to_csv(csv_path, index=False)

    
```

```python
for dataset in ["MDA", "NU"]:
    print("running: " + dataset)
    heatmap_parent = "/Volumes/rsrch5/home/trans_mol_path/cercan/BE_master/2.intermediate/2.0_selectedModel/250313/36138437_24_f16/heatmap_05window/{}/".format(dataset)
    HEATMAP_OUTPUT_path = os.path.join(heatmap_parent,"raw/HEATMAP_OUTPUT/")
    csv_folder = os.path.join(heatmap_parent, "csv_250417")
    os.makedirs(csv_folder , exist_ok= True)
    flow_groups = [f for f in os.listdir(HEATMAP_OUTPUT_path) if os.path.isdir(os.path.join(HEATMAP_OUTPUT_path, f))]

    for flow_group in flow_groups:
        sample_list = [f for f in os.listdir(os.path.join(HEATMAP_OUTPUT_path, flow_group)) if os.path.isdir(os.path.join(HEATMAP_OUTPUT_path, flow_group, f))]

        for sample in sample_list:
            blockmap_path = os.path.join(HEATMAP_OUTPUT_path, flow_group, sample,sample + '_224px_blockmap.h5')
            csv_path = os.path.join(csv_folder, sample + '.csv')
            
            h52csv(blockmap_path, csv_path)
```
