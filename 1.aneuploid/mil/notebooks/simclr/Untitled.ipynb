```python
import os
import openslide
```

```python
def get_tif_files(folder_path):
    """Get a list of all TIFF files in the specified folder."""
    tif_files = []
    for file_name in os.listdir(folder_path):
        if file_name.lower().endswith('.tif') or file_name.lower().endswith('.tiff'):
            tif_files.append(os.path.join(folder_path, file_name))
    return tif_files

def print_tif_dimensions(folder_path):
    """Print the dimensions of all TIFF files in the specified folder if any dimension is smaller than 1024."""
    tif_files = get_tif_files(folder_path)
    for tif_file in tif_files:
        try:
            slide = openslide.OpenSlide(tif_file)
            width, height = slide.dimensions
            if width < 1024 or height < 1024:
                print(f"File: {tif_file}")
                print(f"Dimensions: {width} x {height}")
            slide.close()
        except Exception as e:
            print(f"Error reading {tif_file}: {e}")

base_path = "/rsrch5/home/trans_mol_path/cercan/BE_master/0.input/flow/roi/flow/MDA12_ROI_pyramid"

print_tif_dimensions(base_path)
```
