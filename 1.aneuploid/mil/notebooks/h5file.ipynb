```python
import os
import h5py
import pandas as pd
import numpy as np
```

```python
def h52csv(h5_path,csv_path):
    with h5py.File(h5_path, 'r') as h5data:
        A_minmax = np.array(h5data['A_minmax'])
        tile_probabilities = np.array(h5data['tile_probabilities'])
        coords = np.array(h5data['coords'])
    
    df = pd.DataFrame({
        'x': coords[:, 0],
        'y': coords[:, 1],
        'A_minmax': A_minmax.flatten(),  # Flatten to ensure it's a 1D array
        'tile_probabilities': tile_probabilities.flatten()
    })
    df.to_csv(csv_path, index=False)

    
```

```python
heatmap_parent = "/rsrch5/home/trans_mol_path/cercan/data/BE/aneuploid/clam/trainings/sep24_progress/1_0/heatmap/16768515_21_s1_fold3_MDA_window05"
HEATMAP_OUTPUT_path = os.path.join(heatmap_parent,"raw/HEATMAP_OUTPUT/")
csv_folder = os.path.join(heatmap_parent, "csv")
os.makedirs(csv_folder , exist_ok= True)
```

```python
for flow_group in ['aneuploid', 'diploid']:
    for subdir, _, _ in os.walk(os.path.join(HEATMAP_OUTPUT_path,flow_group)):
        base_wsi_name= os.path.basename(subdir)
        blockmap_path = os.path.join(subdir , base_wsi_name+'_blockmap05window.h5')
        if os.path.isfile(blockmap_path): 
            csv_path = os.path.join(csv_folder,base_wsi_name+'.csv')
            print("running: "+base_wsi_name)
            h52csv(blockmap_path,csv_path)
```
