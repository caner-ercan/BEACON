```python
import os
import h5py
import pandas as pd
import numpy as np
```

```python
def min_max_scaling(h5_path):
    with h5py.File(h5_path, 'r') as h5data:
        # Read the specified dataset into a NumPy array
        np_data = np.array(h5data['attention_scores'])
    min_val = np.min(np_data)
    max_val = np.max(np_data)
    scaled_data = (np_data - min_val) / (max_val - min_val)
    with h5py.File(h5_path, 'a') as h5data:
        h5data.create_dataset('A_minmax', data=scaled_data)
    return scaled_data

```
