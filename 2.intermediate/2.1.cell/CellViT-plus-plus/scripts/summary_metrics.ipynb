```python
import pandas as pd
import json
import yaml
import os
from pathlib import Path
```

```python
output_path = "/rsrch5/home/trans_mol_path/cercan/data/BE/cell_detect/cellvit/250502_finetuning/logs_local/250505"
sweeps = [f for f in os.listdir(output_path) if f.startswith("sweep")]
results = pd.DataFrame()

for sweep in sweeps:
    sweep_path = os.path.join(output_path, sweep)
    runs = os.listdir(sweep_path)
    for run in runs:
        run_path = os.path.join(sweep_path, run)
        config_path = os.path.join(run_path, "config.yaml")
        scores_path = os.path.join(run_path, "val_results", "scores.json")

        if os.path.exists(config_path) and os.path.exists(scores_path):
            # Read metrics
            with open(scores_path, 'r') as file:
                json_data = json.load(file)
            metric_df = pd.DataFrame([json_data])

            # Read model configs
            with open(config_path, 'r') as file:
                yaml_dict = yaml.safe_load(file)
            checkpoint_path = yaml_dict['checkpoint_path']
            base_name = os.path.splitext(os.path.basename(checkpoint_path))[0]
            hidden_dim = yaml_dict['model']['hidden_dim']
            model_df = pd.DataFrame({'Base_Name': [base_name], 'Hidden_Dim': [hidden_dim]})

            # Combine model configs and metrics
            run_df = pd.concat([model_df, metric_df], axis=1)
            results = pd.concat([results, run_df], ignore_index=True)
        else:
            print(f"Missing file in run: {run_path}")

# Display the results DataFrame
results.to_csv(os.path.join(output_path,'results.csv'), index=False)
```
