```python
import os
import yaml
import shutil
```

```python
output_folder = "/rsrch5/home/trans_mol_path/cercan/data/BE/cell_detect/cellvit/250501_finetuning"
# config_templete_folder = os.path.join(output_folder, "templetes")

checkpoint_folder = "/rsrch5/home/trans_mol_path/cercan/code/CellViT-plus-plus/checkpoints/classifier/sam-h"
classifier_list = [os.path.splitext(f)[0] for f in os.listdir(checkpoint_folder)]

templete_config_path = os.path.join(output_folder, "config_templete.yaml")

with open(templete_config_path, 'r') as file:
    configs = yaml.safe_load(file)
```

```python
for classifier in classifier_list:
    exp_folder = os.path.join(output_folder, classifier)
    os.makedirs(exp_folder, exist_ok=True)

    # Edit values
    log_comment = "cellvit" + classifier
    wandb_dir = os.path.join(exp_folder, "logs_local")
    log_dir = os.path.join(exp_folder, "logs_local")

    classifier_path = os.path.join(checkpoint_folder, classifier + '.pth')

    configs['logging']['wandb_dir'] = wandb_dir
    configs['logging']['log_dir'] = log_dir
    configs['logging']['log_comment'] = f"cellvit_{classifier}"
    configs['checkpoint_path'] = classifier_path
    os.makedirs(wandb_dir, exist_ok=True)

    # Dump template
    export_path = os.path.join(exp_folder, "config_sweep.yaml")
    with open(export_path, 'w') as file:
        yaml.dump(configs, file)

    shutil.copy(os.path.join(output_folder, "label_map.yaml"), os.path.join(exp_folder, "label_map.yaml"))
```
