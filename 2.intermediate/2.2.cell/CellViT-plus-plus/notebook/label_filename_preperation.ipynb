```python
import os
import pandas as pd
import shutil
```

```python
label_map = {
    "Lymphocyte":0,
    "Plasma":1,
    "OtherImmune":2,
    "Epithelial":3,
    "Stroma":4}
```

```python
qp_proj_path ="/rsrch5/home/trans_mol_path/cercan/BE_master/2.intermediate/2.2.cell/training/qp_proj"
annotations_path = os.path.join(qp_proj_path,"finetune_annotations")
csv_folder= os.path.join(annotations_path,"corrected_point")
roi_folder= os.path.join(annotations_path,"tiles")

split_table_path = "/rsrch5/home/trans_mol_path/cercan/BE_master/2.intermediate/2.0.input/subset_list.csv"
output= os.path.join(qp_proj_path,"cellvit_finetune_annotations")

```

```python
def csv_converter(csv_path, label_map):
    csv = pd.read_csv(csv_path, sep=",")
    csv.iloc[:,0:2] = csv.iloc[:,0:2].astype(int)
    csv.iloc[:,-1] = csv.iloc[:,-1].map(label_map)
    return csv
    
```

```python
train_image_path= os.path.join(output,"train","images")
train_csv_path= os.path.join(output,"train","labels")

test_image_path= os.path.join(output,"test","images")
test_csv_path= os.path.join(output,"test","labels")

os.makedirs(train_image_path, exist_ok=True)
os.makedirs(train_csv_path, exist_ok=True)
os.makedirs(test_image_path, exist_ok=True)
os.makedirs(test_csv_path, exist_ok=True)

```

```python
split_table = pd.read_csv(split_table_path, sep=",")

file_basenames = [os.path.splitext(f)[0] for f in os.listdir(roi_folder)]

train_split =[]
val_split = []
for filebasename in file_basenames:
    #remove commas in the filename
    filebasename_new = filebasename.replace(",", "_")

    csv_path = os.path.join(csv_folder, filebasename + ".csv")
    labels = csv_converter(csv_path, label_map)
    

    wsi_name = filebasename[:5]
    split = split_table.loc[split_table["slide_id"] == wsi_name]["subset"].values[0]
    if split == "test":
        shutil.copy(os.path.join(roi_folder, filebasename + ".png"), os.path.join(test_image_path, filebasename_new + ".png"))
        labels.to_csv(os.path.join(test_csv_path, filebasename_new + ".csv"), index=False,header=False)
    elif split == "train":
        shutil.copy(os.path.join(roi_folder, filebasename + ".png"), os.path.join(train_image_path, filebasename_new + ".png"))
        labels.to_csv(os.path.join(train_csv_path, filebasename_new + ".csv"), index=False,header=False)
        train_split.append(filebasename_new)
    elif split == "val":
        shutil.copy(os.path.join(roi_folder, filebasename + ".png"), os.path.join(train_image_path, filebasename_new + ".png"))
        labels.to_csv(os.path.join(train_csv_path, filebasename_new + ".csv"), index=False,header=False)
        val_split.append(filebasename_new)
    else:
        print(wsi_name +"not in the list")

# save train_split and val_split as csv to os.path.join(output,"splits", "XXX.csv")

os.makedirs(os.path.join(output, "splits"), exist_ok=True)
train_split_df = pd.DataFrame(train_split)
val_split_df = pd.DataFrame(val_split)
train_split_df.to_csv(os.path.join(output, "splits", "train.csv"), index=False,header=False)
val_split_df.to_csv(os.path.join(output, "splits", "val.csv"), index=False,header=False)
```
